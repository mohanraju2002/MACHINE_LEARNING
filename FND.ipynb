```python
import pandas as pd

df = pd.read_csv("fake_news_dataset_1200.csv")

# Convert text to string
df["text"] = df["text"].astype(str)

# Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(df["text"])

# Labels (IMPORTANT FIX)
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(df["label"])   # ✅ use label column

# Split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)

# Accuracy
from sklearn.metrics import accuracy_score
pred = model.predict(x_test)
print("Accuracy:", accuracy_score(y_test, pred))

# Prediction 1
new_text1 = ["Breaking: Scientists discovered water on Mars"]
vec1 = vectorizer.transform(new_text1)
pred1 = model.predict(vec1)
print("Prediction 1:", le.inverse_transform(pred1))   # ➜ real/fake

# Prediction 2
new_text2 = ["Humans can survive without food for years"]
vec2 = vectorizer.transform(new_text2)
pred2 = model.predict(vec2)
print("Prediction 2:", le.inverse_transform(pred2))   # ➜ real/fake
```
